mfc means Marker Face Color

It controls the inside/fill color of the marker                                                           
marker='o'	Marker shape = circle
mfc='red'	Marker face/fill color = red
mec='blue'	Marker edge/border color = blue
ms=10	Marker size
alpha=0.5	Transparency

```python
plt.figure(figsize=(12,3))
plt.plot( df['Days'] , df['Sales'],marker='o',mfc='red' ,ms=10)
plt.show()
```

```python
plt.figure(figsize=(10,3))
plt.scatter(tip,bill,color='red')
plt.xlabel("Tips USD$")
plt.ylabel("Bill USD$")
plt.title("Bill VS Tips Distribution")
plt.show()
```

```python
colors=[]
for i in range(len(bill)):
    amt=bill[i]*0.01
    if amt<tip[i]:
        colors.append('green')
    else:
        colors.append('hotpink')
```

```python
plt.figure(figsize=(10,3))
plt.scatter(tip,bill,color=colors)
plt.xlabel("Tips USD$")
plt.ylabel("Bill USD$")
plt.title("Bill VS Tips Distribution")
plt.show()
```

autopct="%.2f%%" ko break karo:
% → percentage formatting start
.2f → number ko 2 decimal places tak dikhayega
%% → actual % symbol print karega
Example

Agar kisi month ki sales total sales ka:

25.6789%

hai, to chart mein dikhega:

25.68%
Simple Hinglish

autopct="%.2f%%" = Pie chart ke andar percentage ko 2 decimal places ke saath show karo.

Agar aap:

autopct="%.1f%%"

likhte ho → 25.7%

Aur:

autopct="%.0f%%"

likhte ho → 26%

👉 Interview mein: autopct is used to display the percentage values on each slice of a pie chart.

```python
plt.Figure(figsize=(6,3))
plt.pie(df['Sales'],labels=df['Month'],autopct="%.2f%%")
plt.show()
```
