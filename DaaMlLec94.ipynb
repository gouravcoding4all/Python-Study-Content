```python
from sklearn.cluster import KMeans
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
df = sns.load_dataset('iris')
```

```python
y=df['species']
```

```python
df=df.drop(columns=['species'])
```

```python
labels = kmeans.fit_predict(df)
```

```python
plt.Figure(figsize=(6,3))
sns.scatterplot(x='petal_length',y='petal_width',data=df)
plt.show()
```

```python
plt.figure(figsize=(12,3))
plt.subplot(1,2,1)
sns.scatterplot(x='sepal_length',y='sepal_width',data=df,hue=y)
plt.subplot(1,2,2)
sns.scatterplot(x='sepal_length',y='sepal_width',data=df,hue=labels)
plt.show()
```

Unsupervised Learning & K-Means
1. Unsupervised Learning kya hoti hai?

File ki starting mein likha hai:

Unsupervised Learning
Clustering => Group

Simple language mein:

Unsupervised Learning = Data mein already answer/label nahi diya hota, model khud pattern find karta hai.

Real-life example

Maan lo ek shopkeeper ke paas 100 customers hain:

Customer	Spending	Visits
A	₹5,000	2
B	₹50,000	15
C	₹6,000	3
D	₹45,000	12

Shopkeeper bolta hai:

"Mujhe customers ko groups mein divide karna hai."

Lekin usne pehle se nahi bataya ki kaunse customer kis group mein hain.

Model khud groups bana sakta hai:

Group 1 → Low spending
Group 2 → Medium spending
Group 3 → High spending

Ye Clustering hai.

Yaad rakho:

Clustering = Similar data ko ek group mein rakhna.

2. K-Means import karna

File mein:

from sklearn.cluster import KMeans

Iska matlab:

sklearn → Machine Learning library
cluster → Clustering ka module
KMeans → K-Means clustering algorithm

Hum Python ko bol rahe hain:

"Mujhe K-Means algorithm use karna hai."

kmeans = KMeans(n_clusters=3)

Yahan:

KMeans

n_clusters=3

"Mere data ko 3 groups mein divide karo."

For example:

Group 0
Group 1
Group 2

Important: Python mein counting usually 0 se start hoti hai.

Isliye 3 clusters ke labels generally:

0, 1, 2

honge.

9. Species ko remove kyun kiya?

Ye Unsupervised Learning ka important concept hai.

Hum K-Means ko actual answer nahi dena chahte.

Agar hum species bhi model ko de dete, to model ko pehle se pata hota:

"Ye setosa hai, ye versicolor hai..."

"Tum khud similar flowers ko groups mein divide karo."

se species ko temporarily save kar liya.

Phir:

se model ke input se hata diya.

10. K-Means actually run karna

Sabse important line:

labels = kmeans.fit_predict(df)

Is line ko achhe se samjho.

fit_predict()

Ismein do kaam hote hain:

fit

Model data ko study karta hai.

Matlab:

"Kaunsa data ek doosre ke similar hai?"

predict

Model har row ko ek cluster number assign karta hai.

Example:

Customer A → 0
Customer B → 1
Customer C → 0
Customer D → 2

To:

labels

mein cluster numbers aa jaate hain.

11. labels kya hai?

File mein labels ka output ek array hai:

array([1, 1, 1, ..., 0, 2, 0, 2, ...])

Iska simple meaning:

Har flower ko K-Means ne ek group/cluster assign kiya.

For example:

Flower 1 → Cluster 1
Flower 2 → Cluster 1
Flower 3 → Cluster 1
Flower 4 → Cluster 0
Flower 5 → Cluster 2
Important:

0, 1, 2 flower species ke naam nahi hain.

Ye sirf cluster IDs hain.
