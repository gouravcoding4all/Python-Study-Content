```python
#print missing value columns names:
for col in df.columns:
    if df[col].isnull().sum()>0:
        print(col,'\t',df[col].isnull().sum(),'\t',df[col].dtype)
```

```python
#fill missing values numerical with mean(average) and categorical/text with mode(most_frequent)
for col in df.columns:
    if df[col].isnull().sum()>0:
        if df[col].dtype in [object,str]:
            print(col,'\t',df[col].isnull().sum(),'\t',df[col].dtype,'\t',df[col].mode()[0])
            df[col]=df[col].fillna(df[col].mode()[0])
        else:
            print(col,'\t',df[col].isnull().sum(),'\t',df[col].dtype,'\t',df[col].mean())
            df[col]=df[col].fillna(round(df[col].mean(),2))
# Column ka average (mean) nikalo
#➡️ Usko 2 decimal places tak round karo
#➡️ Jahan NaN hai, wahan woh average bhar do.

#🧠 Short trick:

#NaN → Mean → Round(2) → Fill
```

```python
#handle missing value numerical column
for col in df.select_dtypes(include=np.number).columns:
    if df[col].isnull().sum() > 0:
        print(col, '\t', df[col].dtype, '\t',
              df[col].isnull().sum(), '\t Mean :',
              round(df[col].mean(), 2))

        df[col] = df[col].fillna(round(df[col].mean(), 2))
```
