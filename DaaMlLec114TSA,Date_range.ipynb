```python
Lecture 114 – Time Series Analysis (TSA)
1. Time Series Analysis kya hota hai?
Notebook mein likha hai:
Train a model on the data which collect over the time.

Simple language mein:

Time Series Analysis ka matlab hai time ke according data ko samajhna aur future prediction karna.

Example: Maan lo ek shop hai. Uski monthly sales kuch aisi hain:
Month	Sales
January	₹10,000
February	₹15,000
March	₹20,000
April	₹25,000
May	₹30,000

Ab hum past sales dekhkar predict kar sakte hain ki June mein kitni sales ho sakti hain.

Real-life examples:
- Weather forecasting
- Stock market prediction
- Monthly company sales
- Daily website visitors
- Electricity consumption
2. Import Libraries
Code:
import pandas as pdimport numpy as np



Chalo ek-ek line samajhte hain.
Line 1: import pandas as pd
- import = Kisi library ko apne code mein lana.
- pandas = Python ki library jo data handle karti hai.
- as pd = Pandas ka short name rakhna.
Example:
import pandas as pdpd.DataFrame()



Socho Pandas ek school ki register book hai, jisme students ka naam, age aur marks store kar sakte hain.
Line 2: import numpy as np
- import = Library ko load karna.
- numpy = Numbers aur mathematical calculations ke liye library.
- as np = NumPy ka short name.
Example:
import numpy as npnp.array([10, 20, 30, 40])



Output:
array([10, 20, 30, 40])



Difference:
Pandas	NumPy
Data handling	Numerical calculations
DataFrame	Array
Table format	Array format

3. Date Range Create Karna
Important Python Command
Your notebook code:
dates = pd.date_range(    start='2024-01-01',    periods=24,    freq="ME")



Is code ko line-by-line samajhte hain.
Line 1: dates =
dates =



Iska matlab hai jo result milega, usko dates naam ke variable mein store kar do.

Example:
a = 10



Ab a ke andar 10 store ho gaya.
Line 2: pd.date_range()
pd.date_range()



Iska use automatically dates generate karne ke liye hota hai.

Jaise aapko January 2024 se lekar December 2025 tak dates chahiye, toh manually likhne ki zarurat nahi hai.

Python automatically bana dega.
Line 3: start='2024-01-01'
start='2024-01-01'



Iska matlab hai ki date generate karna kahan se start karna hai.

Yahan:
- 2024 = Year
- 01 = January
- 01 = First day
Matlab starting date 1 January 2024 hai.
Line 4: periods=24
periods=24



Iska matlab hai total kitni dates generate karni hain.

Yahan 24 ka matlab hai 24 dates.

Dhyan dena: Ye 24 days nahi hain. Kyunki humne frequency monthly rakhi hai, isliye 24 months generate honge.
Line 5: freq="ME"
freq="ME"



Ye batata hai ki date kitne interval ke baad generate hogi.

ME ka full form hai Month End.

Matlab har month ki last date generate hogi.

Example:
- January → 31 January
- February → 29 February (2024 leap year hai)
- March → 31 March
- April → 30 April
Complete code ek baar phir dekho:
dates = pd.date_range(    start='2024-01-01',    periods=24,    freq="ME")print(dates)



Expected output ki starting dates:
2024-01-31
2024-02-29
2024-03-31
2024-04-30
2024-05-31
...
2025-12-31


4. dates ko print karna
Notebook code:
dates



Iska matlab hai dates variable ke andar jo bhi data store hai, usko display karo.

Jupyter Notebook mein sirf variable ka naam likhne se bhi output show ho jata hai.

Python file mein aap ye use kar sakte ho:
print(dates)




5. Trend







savemyexams.com


Notebook definition:
A long-term movement upward or downward is trend.

Simple Hinglish:

Trend ka matlab hai data ka long-term direction.

Data overall upar ja raha hai ya neeche ja raha hai.
Example 1: Upward Trend
Maan lo ek company ki sales:
Year	Sales
2021	10,000
2022	20,000
2023	30,000
2024	40,000

Sales continuously increase ho rahi hain.

Isko bolenge Upward Trend.
Example 2: Downward Trend
Year	Sales
2021	50,000
2022	40,000
2023	30,000
2024	20,000

Sales continuously decrease ho rahi hain.

Isko bolenge Downward Trend.

Interview answer:

Trend represents the long-term upward or downward movement in time series data.
6. Seasonality







databites.tech


Notebook definition:
A known period of time for downward or upward is seasonality.

Simple Hinglish:

Seasonality ka matlab hai kisi particular time period mein data ka baar-baar increase ya decrease hona.

Example:

Maan lo ek AC company hai.
- March: AC sales increase
- April: AC sales increase
- May: AC sales increase
- June: AC sales increase
- Winter: AC sales decrease
Agar ye pattern har saal repeat hota hai, toh ise Seasonality kahenge.

Real-life examples:
- Diwali par shopping increase hona.
- Summer mein cold drinks ki sales increase hona.
- Winter mein heaters ki sales increase hona.
- December mein tourism demand increase hona.
Interview answer:

Seasonality refers to a repeating pattern in time series data over a fixed and known period.
7. Noise







biz.libretexts.org


Notebook definition:
A sudden big change in the movement.

Simple Hinglish:

Noise ka matlab hai data mein random aur unpredictable changes hona.

Example:

Ek shop ki daily sales:
Monday      1000
Tuesday     1100
Wednesday    950
Thursday    1050
Friday      5000
Saturday     900
Sunday      1000


Friday ko achanak sales ₹5,000 ho gayi.

Agar iska koi regular pattern nahi hai, toh ye unusual random fluctuation ho sakta hai.

Technically, noise ka matlab random variation hota hai. Har sudden big change necessarily noise nahi hota.

Interview answer:

Noise refers to random and unpredictable fluctuations in time series data that cannot be explained by trend or seasonality.

8. Important Difference – Trend vs Seasonality vs Noise
Concept	Simple meaning	Example
Trend	Long-term direction	Sales har saal badhna
Seasonality	Repeating pattern	Summer mein AC sales badhna
Noise	Random fluctuation	Achanak sales change hona

9. Test Yourself – Practice Questions
Ab dekhte hain ki aapko kitna samajh aaya.
Lecture 114 – Mini Quiz
5/5 attempted

Q1. A company ki sales har saal increase ho rahi hain. Ye kya hai?

Noise

Trend

Seasonality

Correct!Long-term increase ko upward trend kehte hain.


Q2. Har Diwali par shopping increase hoti hai. Ye kya hai?

Seasonality

Noise

Downward Trend

Correct!Diwali ka repeating yearly pattern seasonality hai.


Q3. pd.date_range() ka use kya hai?

Dates generate karna

Image banana

Numbers delete karna

Correct!pd.date_range() automatically date sequence banata hai.


Q4. periods=24 aur freq='ME' ka kya meaning hai?

24 days

24 months

24 years

Correct!ME means Month End, isliye 24 monthly dates generate hongi.


Q5. NumPy ko short name se kaise likhte hain?

pd

np

py

Correct!import numpy as np

Your score: 5/5
Try Again
My suggestion: Is lecture mein sabse important command pd.date_range() hai. Iske baad Trend, Seasonality aur Noise ke differences ko achhe se samajhna, kyunki ye Time Series Analysis ke basic concepts hain.

Next step mein hum isi notebook ko continue karke actual time series dataset, visualization aur forecasting practically seekh sakte hain.
```
