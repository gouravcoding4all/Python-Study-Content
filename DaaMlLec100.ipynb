```python
1. Unsupervised Learning & Feature Engineering

File ki starting mein likha hai:

Unsupervised Learning

Simple language mein:

Machine Learning mein kabhi-kabhi hamare paas data hota hai, lekin answer/target nahi hota.

Example:

Height	Weight
150	45
160	55
180	80

Yahan machine ko koi answer nahi diya gaya ki kaunse group mein hai.

Machine khud data ke andar pattern/group dhundhti hai.

Isko Unsupervised Learning kehte hain.

2. Feature Engineering

File mein hai:

Feature Engineering

Feature ka simple meaning:

Feature = Data ka column

Example:

Age
Salary
Height
Weight
Experience

Ye sab features hain.

Feature Engineering kya hai?

Raw data ko aise modify karna ki Machine Learning model ke liye data aur useful ho jaye.

File mein do important methods diye gaye hain:

1. Feature Selection
2. Feature Extraction (Dimension Reduction)
3. Feature Selection

Suppose hamare paas 10 columns hain:

Age
Salary
Height
Weight
City
Experience
Gender
Marks
Phone
ID

Har column useful nahi hota.

Hum kuch useful columns choose kar sakte hain:

Age
Salary
Experience
Marks

Baaki columns hata dete hain.

Isko:

Feature Selection

kehte hain.

Simple example

Socho tumhare school bag mein 20 books hain, lekin class mein sirf 5 books chahiye.

Tum sirf required 5 books rakhoge.

Ye Feature Selection jaisa hai.

4. Feature Extraction / Dimension Reduction

File mein likha hai:

Feature Extraction (Dimension Reduction)

Yahan hum multiple columns ki information ko kam number ke new columns mein convert karne ki koshish karte hain.

Example:

Column 1
Column 2
Column 3
Column 4

ko reduce karke:

New Column 1
New Column 2

kar sakte hain.

Important technique:

PCA

5. Dimensional Reduction

File mein explanation diya hai:

Dimensional Reduction is the process to merge the columns together to reduce the no of columns...

Simple meaning:

Dimensional Reduction = columns ki number ko kam karna.

Example:

Pehle:

4 columns

PCA ke baad:

2 columns

Ho gaye.

Tumhari file mein exactly yehi hua hai:

Before PCA → 4 columns
After PCA  → 2 columns
Kyun?

Bahut zyada columns hone par model complex ho sakta hai.

Isliye kabhi-kabhi hum information ko kam dimensions mein represent karte hain.

6. Libraries Import Karna

Code:

import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

Ab ek-ek line samjho.

Line 1
import seaborn as sns
import

Python ko bol rahe hain:

"Mujhe Seaborn library use karni hai."

seaborn

Seaborn ka use mainly graphs/charts banane ke liye hota hai.

as sns

Hum Seaborn ka short name bana rahe hain:

seaborn → sns

Isliye baad mein:

sns.load_dataset()

likh sakte hain.

7. Matplotlib
import matplotlib.pyplot as plt

Matplotlib bhi graphs banane ki library hai.

Humne iska short name rakha:

plt

Isliye:

plt.scatter()
plt.show()

likhte hain.

Simple example
plt

ko tum ek drawing tool samajh sakte ho.

8. Pandas
import pandas as pd

Pandas ka use tables/data ke saath kaam karne ke liye hota hai.

Example:

Name    Age    Salary
A       20     30000
B       25     40000

Ye type ka data Pandas easily handle karta hai.

Short name:

pandas → pd
9. NumPy
import numpy as np

NumPy numerical calculations aur arrays ke liye useful hai.

Short name:

numpy → np
10. Iris Dataset Load Karna

Code:

df=sns.load_dataset('iris')

Is line ko bahut simple way mein samjho.

sns

Seaborn.

.load_dataset()

Dataset load karo.

'iris'

Iris naam ka famous dataset.

df

Dataset ko humne variable df mein store kiya.

So:

df = iris dataset
11. Iris Dataset Kya Hai?

Iris flowers ka data hai.

Ismein flower ke measurements hain.

Main columns:

sepal_length
sepal_width
petal_length
petal_width
species

Example:

5.1  3.5  1.4  0.2  setosa

Yani flower ke measurements aur uski species.

12. df.head(2)

Code:

df.head(2)

head() ka matlab:

Dataset ki starting rows dikhao.

2 ka matlab:

Starting ki 2 rows dikhao.

Output mein:

sepal_length  sepal_width  petal_length  petal_width  species
5.1           3.5          1.4           0.2          setosa
4.9           3.0          1.4           0.2          setosa
Simple example

Agar notebook mein 1000 students hain aur tum sirf first 5 dekhna chahte ho:

df.head(5)
13. X aur y Banana

Code:

X=df.drop(columns=['species'])
y=df['species']

Ye Machine Learning ka bahut important concept hai.

First line
X=df.drop(columns=['species'])

Hum species column ko hata rahe hain.

Kyun?

Kyuki species hamara answer/target hai.

To X mein sirf:

sepal_length
sepal_width
petal_length
petal_width

bachenge.

X = Input

Machine ko ye information di jayegi.

14. y=df['species']
y=df['species']

Yahan hum sirf species column le rahe hain.

y = Target / Answer

Example:

X → flower measurements
y → flower species

Machine ko:

X → input
y → answer

samajhna hai.

15. X.head()

Code:

X.head()

Ye X ki first 5 rows dikhata hai.

Ab species nahi hai.

Output:

sepal_length
sepal_width
petal_length
petal_width
16. StandardScaler Import

Code:

from sklearn.preprocessing import StandardScaler

Yahan hum Scikit-learn se:

StandardScaler

le rahe hain.

StandardScaler kya karta hai?

Different scales ko approximately same scale par laata hai.

Example:

Age       → 20, 30, 40
Salary    → 30000, 50000, 80000

Salary ke numbers bahut bade hain.

StandardScaler data ko standard scale par convert karta hai.

17. StandardScaler Apply Karna

Code:

X = StandardScaler().fit_transform(X)

Ye important line hai.

Isko 3 parts mein samjho:

StandardScaler()

Scaler object banaya.

.fit()

Data se mean aur standard deviation jaisi information learn karta hai.

.transform()

Data ko scale karta hai.

Aur:

fit_transform()

dono ka kaam ek saath karta hai.

18. Example of Scaling

Suppose:

Age = 20

Scaling ke baad ye ho sakta hai:

-1.2

Another value:

Age = 40

ho sakta hai:

1.1

Iska matlab ye nahi hai ki person's age negative ho gayi.

Sirf mathematical scale change hua hai.

19. X[:2]

Code:

X[:2]

Meaning:

X ki first 2 rows dikhao.

Output:

[[-0.90068117,  1.01900435, -1.34022653, -1.3154443 ],
 [-1.14301691, -0.13197948, -1.34022653, -1.3154443 ]]

Pehle X mein 4 columns the.

Ab bhi:

4 columns

hain.

Bas values scale ho gayi hain.

20. PCA Import

Code:

from sklearn.decomposition import PCA

Yahan Scikit-learn se:

PCA

import kiya.

PCA ka full form:

Principal Component Analysis

PCA ka main kaam:

Bahut saare columns ko kam dimensions mein convert karna while retaining as much information as possible.

Tumhari file mein:

4 columns → 2 columns
21. PCA Object Banana

Code:

pca=PCA(n_components=2)

Yahan hum PCA ko bol rahe hain:

"Mujhe final mein 2 columns chahiye."

n_components=2

Meaning:

Original dimensions = 4
New dimensions = 2
22. Before PCA

Code:

#Before PCA
X.shape

# se start hone wali line comment hoti hai.

Python isko execute nahi karta.

Comment sirf human ko samjhane ke liye hai.

X.shape

dataset ka shape batata hai.

Output:

(150, 4)

Iska matlab:

150 rows
4 columns
23. X.shape ko School Example se Samjho

Imagine classroom mein:

150 students

Aur har student ke paas:

4 subjects

To:

(150, 4)

Matlab:

150 students × 4 subjects
24. PCA Apply Karna

Code:

X=pca.fit_transform(X)

Ye PCA ka actual kaam karta hai.

fit

PCA data ko study karta hai.

Ye dekhta hai:

"Data mein sabse important information kis direction mein hai?"

transform

Phir original 4 columns ko naye PCA columns mein convert karta hai.

fit_transform

Dono kaam ek saath.

25. After PCA

Code:

#After PCA
X.shape

Output:

(150, 2)

Meaning:

150 rows
2 columns

Pehle:

150 × 4

Ab:

150 × 2
Main achievement:
4 dimensions
      ↓
    PCA
      ↓
2 dimensions
26. PCA ke Baad X

Code:

X[:3]

Output:

[[-2.26470281,  0.4800266 ],
 [-2.08096115, -0.67413356],
 [-2.36422905, -0.34190802]]

Ab har row mein sirf 2 values hain.

Ye:

PC1
PC2

ko represent karti hain.

PC ka matlab:

Principal Component

27. PC1 aur PC2 kya hain?

Ye original columns ke simple direct columns nahi hain.

Original:

sepal_length
sepal_width
petal_length
petal_width

PCA in sabki information ko mathematically combine karke naye dimensions banata hai:

PC1
PC2

Simple example:

Imagine 4 teachers tumhare marks de rahe hain:

Math
Science
English
Computer

PCA unki information ko combine karke ek overall pattern bana sakta hai.

For example:

PC1 = overall academic pattern
PC2 = another important pattern

Exact meaning PCA ke mathematical components/loadings par depend karta hai.

28. pca.explained_variance_ratio_

Code:

pca.explained_variance_ratio_

Ye PCA ka bahut important concept hai.

Output:

[0.72962445, 0.22850762]

Iska meaning:

PC1 → 72.96%
PC2 → 22.85%

approximately.

29. Explained Variance ko 5th-Class Example

Imagine tumhare paas ek bada treasure box hai.

Us treasure ki total important information:

100%

PCA bolta hai:

PC1
72.96%

information ko explain kar raha hai.

PC2
22.85%

information explain kar raha hai.

Total:

72.96 + 22.85
= 95.81%

Yani 2 components milkar approximately:

95.8% information explain kar rahe hain.

30. File ka Calculation

Code:

0.72 + 0.22

Output:

0.94

Matlab approximately:

94%

File mein values rounded karke calculate ki gayi hain.

Actual values:

0.72962445
+
0.22850762

≈

0.9581

ya approximately:

95.81%

So exact explained_variance_ratio_ values ko use karoge to result ~95.8% aata hai; notebook ka 0.94 calculation rounded 0.72 + 0.22 se aaya hai.

31. Last Graph

Code:

plt.figure(figsize=(6,3))

Ye graph ka size set karta hai.

6 → width
3 → height

Matlab graph:

6 units wide
3 units tall
32. Scatter Plot

Code:

plt.scatter(X[:,0],X[:,1])

Ye graph mein dots banata hai.

plt.scatter()

Scatter plot banata hai.

Scatter plot = dots ka graph.

33. X[:,0]

Isko carefully samjho.

X[:,0]

Meaning:

:  → saari rows
0 → first column

So:

X[:,0]

means:

"X ki har row se first column ki value lo."

Ye graph ka:

X-axis

banega.

34. X[:,1]
X[:,1]

Meaning:

: → saari rows
1 → second column

So ye:

"Har row se second column ki value lo."

Ye graph ka:

Y-axis

banega.

35. Complete Scatter Plot
plt.scatter(X[:,0],X[:,1])

Simple language:

PC1 ko X-axis par rakho aur PC2 ko Y-axis par rakho, phir har flower ko ek dot ke form mein show karo.

Graph kuch aisa conceptually hai:

             PC2
              ↑
       •  •
    •     •
  •
          •       •
              •
────────────────────────→ PC1

Har dot ek flower/observation ko represent karta hai.

36. plt.show()

Code:

plt.show()

Meaning:

Graph ko screen par dikhao.

Agar ye line nahi likhenge, kuch environments mein graph display na ho ya behavior different ho sakta hai.

37. Puri File ka Flow

Ab poori notebook ko ek story ki tarah samjho:

Step 1 — Dataset
df=sns.load_dataset('iris')

Iris flower ka data liya.

↓

Step 2 — Input aur Output
X=df.drop(columns=['species'])
y=df['species']

Input:

4 flower measurements

Output:

species

↓

Step 3 — Scaling
X = StandardScaler().fit_transform(X)

Sab features ko comparable scale par laya.

↓

Step 4 — PCA
pca=PCA(n_components=2)

Bola:

Mujhe 2 dimensions chahiye.

↓

Step 5 — Transformation
X=pca.fit_transform(X)
4 columns
    ↓
   PCA
    ↓
2 columns

↓

Step 6 — Information Check
pca.explained_variance_ratio_

Dekha ki new 2 components kitni information explain kar rahe hain.

Approximately:

PC1 → 72.96%
PC2 → 22.85%

Total → 95.81%

↓

Step 7 — Visualization
plt.scatter(X[:,0],X[:,1])

2D graph banaya.

38. Sabse Important Concept: PCA

Agar interview mein koi pooche:

"What is PCA?"

Simple answer:

PCA is a dimensionality reduction technique that converts a large number of features into a smaller number of principal components while preserving as much important information as possible.

Hinglish mein:

PCA ek technique hai jo multiple columns/features ko kam number ke new components mein convert karti hai, taaki dimensions kam ho jayein aur important information maximum possible preserve rahe.

39. Ekdum 5th-Class Example

Socho tumhare paas ek student ki information hai:

Math
Science
English
Computer
Sports
Drawing
Music

Total:

7 features

Ab PCA kehta hai:

"Main in 7 subjects ke patterns ko 2 important dimensions mein represent karunga."

To:

7 Features
    ↓
   PCA
    ↓
2 Components

Ab graph banana easy hai:

PC2 ↑
    |
    |       • •
    |    •
    |  •       •
    |________________→ PC1
40. Ek Important Correction

Notebook ke starting comment mein likha hai:

"merge the columns together to reduce the no of columns, so that my model would not be overfit."

Isko beginner level par aise samajhna theek hai, lekin PCA ka purpose sirf overfitting prevent karna nahi hai.

PCA ke important purposes hain:

dimensionality reduce karna
data ko visualize karna
redundant/correlated information ko compact representation mein lana
computation ko reduce karna
useful variance ko retain karte hue fewer dimensions mein data represent karna

Overfitting kabhi-kabhi reduce ho sakta hai, lekin ye guaranteed result nahi hai.

41. Exam/Interview ke liye Important Lines

Tumhari file se ye lines specially yaad rakho:

df=sns.load_dataset('iris')

Iris dataset load karta hai.

X=df.drop(columns=['species'])

Target column ko X se remove karta hai.

y=df['species']

Target/output store karta hai.

X = StandardScaler().fit_transform(X)

Features ko standard scale par convert karta hai.

pca=PCA(n_components=2)

PCA ko 2 components banane ke liye set karta hai.

X=pca.fit_transform(X)

Original 4 features ko 2 principal components mein transform karta hai.

pca.explained_variance_ratio_

Har principal component kitni variance/information explain karta hai, batata hai.

X.shape

Rows aur columns ki count batata hai.

X[:,0]

First column ki all rows.

X[:,1]

Second column ki all rows.

plt.scatter(X[:,0],X[:,1])

PC1 vs PC2 ka scatter plot banata hai.

🔥 One-line memory trick

PCA ko yaad rakho:

"Bahut saare columns → PCA → kam columns, but maximum possible important information retain."

Aur tumhari notebook mein:

150 rows × 4 features
          ↓
     StandardScaler
          ↓
        PCA
          ↓
150 rows × 2 components
          ↓
      Scatter Plot

Yahi DaaMlLec100(2).ipynb ka complete core flow hai.
```
