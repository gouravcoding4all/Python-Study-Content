Age column mein kitne % values missing (NaN) hain, ye calculate karta hai.

isnull().sum() → Missing values count
df.shape[0] → Total rows
× 100 → Percentage

Example: 10 missing out of 100 rows → 10% missing values.

```python
df.age.isnull().sum()/df.shape[0]*100
```

```python
df.age.isnull().sum()
```

df.shape[0] * 100

👉 df.shape[0] = DataFrame ki total rows

Agar df.shape[0] = 10 → 10 × 100 = 1000

```python
df.shape[0]*100
```

Missing (NaN) values ko previous row ki value se fill karta hai.

Example: 10, NaN, 30 → 10, 10, 30

```python
df.select_dtypes(include=np.number).columns
```

Ye code saare numerical columns ke boxplots ek saath banata hai, jisse outliers identify kar sakte hain.

i=1 → Plot position start
select_dtypes(np.number) → Numerical columns select
subplot(2,3,i) → 2×3 grid mein plots
sns.boxplot() → Boxplot banata hai
i=i+1 → Next position
plt.show() → Plots show karta hai.

```python
i=1
plt.Figure(figsize=(12,6))
for col in df.select_dtypes(include=np.number).columns:
    plt.subplot(2,3,i)
    i=i+1
    sns.boxplot(x=col,data=df)
plt.show()
```
