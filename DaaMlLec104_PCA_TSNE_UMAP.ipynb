```python
# Keep species separately for visualization
df = sns.load_dataset('iris')
y = df['species']
X = df.drop(columns=['species'])
```

```python
from sklearn.preprocessing import StandardScaler

# Scale features before dimensionality reduction
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print("Original shape:", X.shape)
print("Scaled shape:", X_scaled.shape)
```

```python
# =========================
# 1. PCA
# =========================
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt

pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

print("PCA shape:", X_pca.shape)
print("Explained variance ratio:", pca.explained_variance_ratio_)
print("Total variance explained:", pca.explained_variance_ratio_.sum())
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(
    x=X_pca[:, 0],
    y=X_pca[:, 1],
    hue=y,
    palette='Set1',
    s=70
)
plt.title("PCA - Iris Dataset")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.legend(title="Species")
plt.show()
```

```python
# =========================
# 2. t-SNE
# =========================
from sklearn.manifold import TSNE

tsne = TSNE(
    n_components=2,
    random_state=42,
    perplexity=30,
    init='pca',
    learning_rate='auto'
)

X_tsne = tsne.fit_transform(X_scaled)

print("t-SNE shape:", X_tsne.shape)
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(
    x=X_tsne[:, 0],
    y=X_tsne[:, 1],
    hue=y,
    palette='Set1',
    s=70
)
plt.title("t-SNE - Iris Dataset")
plt.xlabel("t-SNE Component 1")
plt.ylabel("t-SNE Component 2")
plt.legend(title="Species")
plt.show()
```

```python
# =========================
# 3. UMAP
# =========================
# If UMAP is not installed, run:
# !pip install umap-learn

from umap import UMAP

umap_model = UMAP(
    n_components=2,
    random_state=42
)

X_umap = umap_model.fit_transform(X_scaled)

print("UMAP shape:", X_umap.shape)
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(
    x=X_umap[:, 0],
    y=X_umap[:, 1],
    hue=y,
    palette='Set1',
    s=70
)
plt.title("UMAP - Iris Dataset")
plt.xlabel("UMAP Component 1")
plt.ylabel("UMAP Component 2")
plt.legend(title="Species")
plt.show()
```

```python
# =========================
# Compare all three methods
# =========================
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

sns.scatterplot(x=X_pca[:, 0], y=X_pca[:, 1], hue=y, palette='Set1', s=60, ax=axes[0], legend=False)
axes[0].set_title("PCA")
axes[0].set_xlabel("PC1")
axes[0].set_ylabel("PC2")

sns.scatterplot(x=X_tsne[:, 0], y=X_tsne[:, 1], hue=y, palette='Set1', s=60, ax=axes[1], legend=False)
axes[1].set_title("t-SNE")
axes[1].set_xlabel("t-SNE 1")
axes[1].set_ylabel("t-SNE 2")

sns.scatterplot(x=X_umap[:, 0], y=X_umap[:, 1], hue=y, palette='Set1', s=60, ax=axes[2])
axes[2].set_title("UMAP")
axes[2].set_xlabel("UMAP 1")
axes[2].set_ylabel("UMAP 2")

plt.tight_layout()
plt.show()
```

```python
1. Problem Kya Hai?

Iris dataset mein 4 features hain:

Sepal Length
Sepal Width
Petal Length
Petal Width

Aur 3 flower categories hain:

Setosa
Versicolor
Virginica

Data ko visualize karna mushkil hota hai jab dimensions zyada ho.

Example

Agar 100 features hain to graph kaise banaoge?

Isliye hum Dimensionality Reduction use karte hain.

2. Dimensionality Reduction Kya Hai?

Simple language mein:

Bahut saare columns ko kam columns mein convert karna bina important information lose kiye.

Example:

4 columns → 2 columns

Taaki graph mein plot kar sake.

3. Dataset Load Karna
df = sns.load_dataset('iris')

Ye Iris dataset load karta hai.

Dataset size:

(150, 5)

Matlab:

150 rows
5 columns
4. Features Aur Target Alag Karna
y = df['species']
X = df.drop(columns=['species'])
X

Input Features

Feature
Sepal Length
Sepal Width
Petal Length
Petal Width
y

Output

Species
Setosa
Versicolor
Virginica
5. Standardization
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
Kyu Zaroori Hai?

Suppose:

Age = 25

Salary = 500000

Salary ka number bahut bada hai.

Model salary ko zyada importance dega.

Isliye sab features ko same scale par laate hain.

StandardScaler ke baad:

Mean ≈ 0

Standard Deviation ≈ 1

6. PCA (Principal Component Analysis)
from sklearn.decomposition import PCA
pca = PCA(n_components=2)
PCA Kya Karta Hai?

Important information ko preserve karte hue dimensions kam karta hai.

4 Features

↓

2 Features

Principal Components

PCA naye columns banata hai:

PC1
PC2

Ye original columns ka combination hote hain.

Example

Original:

A	B	C	D

PCA ke baad:

PC1	PC2
PCA Advantages

✅ Fast

✅ Less Memory

✅ Noise Reduction

✅ Large Dataset Handle Kar Sakta Hai

PCA Disadvantages

❌ Non-linear relationships ko achhe se nahi pakadta

❌ Visualization kabhi-kabhi weak hoti hai

7. PCA Graph

Notebook mein:

sns.scatterplot(...)

Graph banaya gaya hai.

Har color ek flower species ko represent karta hai.

Agar clusters alag-alag dikh rahe hain to PCA achha kaam kar raha hai.

8. t-SNE

Import:

from sklearn.manifold import TSNE
Full Form

t-Distributed Stochastic Neighbor Embedding

Naam bada hai lekin concept simple hai.

t-SNE Kya Karta Hai?

Nearby points ko nearby hi rakhta hai.

Matlab:

Agar original data mein do points paas hain

↓

Graph mein bhi paas rahenge

Parameters
TSNE(
    n_components=2,
    perplexity=30,
    random_state=42
)
Perplexity

Ye decide karta hai kitne neighboring points consider karne hain.

Common values:

5
10
20
30
50

Most common:

30
t-SNE Advantages

✅ Amazing visualization

✅ Hidden clusters dikhata hai

✅ Non-linear patterns samajhta hai

t-SNE Disadvantages

❌ Slow

❌ Large datasets par expensive

❌ New data par directly apply nahi kar sakte

9. t-SNE Graph

Aapko usually PCA se better cluster separation milegi.

Flowers ke groups zyada clearly alag dikhte hain.

10. UMAP

Import:

from umap import UMAP
Full Form

Uniform Manifold Approximation and Projection

UMAP Kya Hai?

Ye PCA aur t-SNE ka combination jaisa samjho.

PCA se zyada powerful

t-SNE se zyada fast

Code
umap_model = UMAP(
    n_components=2,
    random_state=42
)
UMAP Advantages

✅ Fast

✅ Better Clustering

✅ Large Dataset Handle

✅ New Data Transform Kar Sakta Hai

✅ Modern ML mein popular

UMAP Disadvantages

❌ PCA jitna simple nahi

❌ Hyperparameters tune karne pad sakte hain

11. PCA vs t-SNE vs UMAP
Feature	PCA	t-SNE	UMAP
Speed	Fast	Slow	Fast
Visualization	Good	Excellent	Excellent
Large Dataset	Good	Poor	Excellent
New Data Transform	Yes	No	Yes
Non-linear Data	No	Yes	Yes
Industry Usage	Very High	Medium	High
Interview Question
Q1. PCA Kya Hai?

Answer:

PCA ek dimensionality reduction technique hai jo data ki important information preserve karte hue features ko kam dimensions mein convert karti hai.

Q2. PCA Kyu Use Karte Hain?

Answer:

Visualization
Noise Reduction
Faster Training
Less Memory
Q3. PCA Se Pehle Scaling Kyu Karte Hain?

Answer:

Different feature scales ki wajah se model biased ho sakta hai. Isliye StandardScaler use karte hain.

Q4. t-SNE Kya Hai?

Answer:

t-SNE ek non-linear dimensionality reduction technique hai jo similar points ko graph mein paas rakhti hai.

Q5. t-SNE Aur PCA Mein Difference?

Answer:

PCA fast hai aur linear technique hai.

t-SNE slow hai lekin visualization better deta hai.

Q6. UMAP Kya Hai?

Answer:

UMAP ek modern dimensionality reduction technique hai jo fast bhi hai aur better visualization bhi deti hai.

Q7. Industry Mein Sabse Zyada Kya Use Hota Hai?

Answer:

Feature Reduction → PCA
Visualization → t-SNE / UMAP
Large ML Projects → UMAP
Easy Memory Trick

PCA = Fastest

t-SNE = Best Visualization

UMAP = Fast + Best Visualization

👉 Exam ya interview mein agar poocha jaye "PCA, t-SNE aur UMAP mein difference batao", to ye 3-line trick yaad rakhna kaafi hai.
```
