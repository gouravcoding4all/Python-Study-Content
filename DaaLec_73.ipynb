        - if data has no relation with another data point/column (replace it with mean / mode).

```python
for col in df.columns:
    print(col,df[col].dtype,df[col].isnull().sum())
```

```python
for col in df.columns:
    if df[col].isnull().sum()>0:
        print(col,df[col].dtype,df[col].isnull().sum())
```

```python
for col in df.columns:
    if df[col].isnull().sum()>0:
        if df[col].dtype in [object,str]:
            print(col,'\t',df[col].dtype,'\t',df[col].isnull().sum(),df[col].mode()[0])
            df[col]=df[col].fillna(df[col].mode()[0])
        else:
            print(col,'\t',df[col].dtype,'\t',df[col].isnull().sum().mean())
            df[col]=df[col].fillna(df[col].mean())
```

for col in df.columns:

➡️ DataFrame ke har column par loop chalao.

if df[col].isnull().sum() > 0:

➡️ Agar column me missing values (NaN) hain.

if df[col].dtype in [object, str]:

➡️ Agar column text (string) type ka hai.

df[col] = df[col].fillna(df[col].mode()[0])

➡️ Missing values ko Mode (sabse zyada aane wali value) se fill karo.

else:

➡️ Warna (numeric column hai).

df[col] = df[col].fillna(df[col].mean())

➡️ Missing values ko Mean (average) se fill karo.

One-line Trick 🚀

Text Column ➜ Mode se Fill

Number Column ➜ Mean se Fill
