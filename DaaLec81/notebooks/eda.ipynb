```python
df.columns
```

```python
for col in df.columns:
    if df[col].isnull().sum()>0:
        if df[col].dtype in [object , str]:
            print(col,'\t',df[col].dtype,'\t',df[col].isnull().sum(),'\tMode :',df[col].mode()[0])
            df[col]=df[col].fillna(df[col].mode()[0])
        else:
            print(col,'\t',df[col].dtype,'\t',df[col].isnull().sum(),'\tMean :',round(df[col].mean(),2))
            df[col]=df[col].fillna(round(df[col].mean(),2))
```
