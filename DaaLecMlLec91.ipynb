```python
from sklearn.preprocessing import LabelEncoder
```

```python
le =LabelEncoder()
```

```python
df['species'] = le.fit_transform(df['species'])
```

```python
#data split inyto X,Y
X = df.drop(columns=['species'])
y = df['species']
```

```python
#Data split into train and test
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.20,random_state=11)
```

Socho 100 students hain aur aap randomly 80 ko training aur 20 ko testing ke liye select kar rahe ho.

Without random_state → har baar students change ho sakte hain.
random_state=11 → ek particular random selection fix ho jayega.
random_state=20 → doosra selection fix ho jayega.

Important: 11 ka koi special meaning nahi hai. Aap 0, 1, 42, 100 etc. bhi use kar sakte ho.

👉 Short definition:
Random state = random splitting ko reproducible/fixed banane ka seed.

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC 
```

```python
models = { 
    "Logistic Regression" : LogisticRegression(),
    "Decision Tree Classifier" : DecisionTreeClassifier(),
    "Random Forest Classifier" : RandomForestClassifier(),
    "K-Neighbors Classifier" : KNeighborsClassifier(),
    "Support Vector Classifier" :SVC()
    }
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix
```

```python
for model_name,model in models.items():
    model.fit(X_train,y_train)
    y_pred = model.predict(X_test)
    print(model_name,'\nAccuracy Score :',round(accuracy_score(y_test,y_pred)*100,2))
    print("\nConfusion Matrix\n",confusion_matrix(y_test,y_pred))
    print("---------------------------------------------------")
```

27. KNN → Lazy Model

Last mein comment hai:

#KNN -> Lazy_Model

KNN ko commonly Lazy Learning Model kaha jata hai.

Why?

KNN training ke time par generally model ko complex rules learn nahi karta.

Instead, prediction ke time:

New data
   ↓
Nearest data points find
   ↓
Unke basis par class decide

👉 Simple: KNN actual decision prediction ke time karta hai, isliye ise Lazy Model kaha jata hai.
