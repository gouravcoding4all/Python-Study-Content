### 1.Support  
Support tells us how frequency an item or item combination appears in all transactions

### Formula
Support(X) = Transaction containing (X) / Total Transaction
Support (Milk) = 4/5 = 0.8 => 80%


### 2.Confidence
Confidence measures <b>how often Y occurs when X Occurs.</b>
### Formula
<b>Confidence(X->Y) = Support(X,Y)/Support(X)</b> <br>
Confidence(Milk -> Bread) = Support(Milk,Bread)/Support(Milk) = 3/4 = 0.75 => 75%

### 3.Lift
### Lift tells us whether X and Y Occur together more often than we would expected by chance
### Formula  
<b>Lift(X -> Y) = Confidence(X->Y)/Support(Y) </b>  
 Lift(Milk->Bread) = 0.75/0.80 => 0.9375

```python
1. Association Rules kya hote hain?
Tumhari file ki first line hai:
###Association Rules:
Iska matlab hai ki ab hum Association Rules naam ka topic padhne wale hain.
Agli line:
Association rule learning is an unsupervised machine learning technique used to discover relationships or patterns between items in a dataset.
Chalo ise chhote parts mein samajhte hain.
- Association: Do ya zyada cheezon ke beech connection ya relationship.
- Rule: Ek niyam ya pattern.
- Learning: Data se kuch seekhna.
- Unsupervised Machine Learning: Machine ko data diya jata hai, lekin har example ka correct answer pehle se label karke nahi diya jata.
- Dataset: Data ka collection, jaise customer ki shopping list.
Real-life example:
Maan lo ek shopkeeper ke paas 1,000 customers ki shopping lists hain. Woh check karna chahta hai ki jo customers Milk kharidte hain, kya woh Bread bhi kharidte hain?
Machine shopping data mein patterns find kar sakti hai.
Example:
- Customer Milk kharidta hai.
- Customer Bread bhi kharidta hai.
- Dono items baar-baar ek saath kharide jaate hain.
Toh machine ek rule discover kar sakti hai:
Milk → Bread
Iska matlab hai ki Milk kharidne wale customers mein Bread kharidne ka bhi pattern ho sakta hai. Lekin iska matlab yeh nahi ki har Milk kharidne wala Bread zaroor kharidega.
Association Rule Learning ka main purpose: Data mein discover karna ki kaunsi cheezein aksar ek saath hoti hain.

2. Suppose We Have These Transactions
Tumhari file mein agla section hai:
###Suppose We have these transactions
Iska simple meaning hai: “Maan lo hamare paas ye shopping transactions hain.”
Transaction ka matlab hai ek baar ki shopping ya purchase record.
Tumhari file mein ye data diya gaya hai:
Transaction	Items (kharidi gayi cheezein)
T1	Milk, Bread
T2	Milk, Bread, Butter
T3	Bread, Butter
T4	Milk, Butter
T5	Milk, Bread, Butter
Ab har row ko samajhte hain.
- T1: Customer ne Milk aur Bread kharida.
- T2: Customer ne Milk, Bread aur Butter kharida.
- T3: Customer ne Bread aur Butter kharida.
- T4: Customer ne Milk aur Butter kharida.
- T5: Customer ne Milk, Bread aur Butter kharida.
Yahan T ka matlab Transaction hai aur T1 se T5 tak total 5 transactions hain.
Important observation
- Milk total 4 transactions mein hai: T1, T2, T4, T5.
- Bread total 4 transactions mein hai: T1, T2, T3, T5.
- Butter total 4 transactions mein hai: T2, T3, T4, T5.
- Milk aur Bread saath mein 3 transactions mein hain: T1, T2, T5.
Ab inhi numbers se hum Support, Confidence aur Lift calculate karenge.

3. Support — Item kitni baar aaya?
Tumhari file ki next line hai:
### 1.Support
Yeh ek heading hai. Iske baad hum Support naam ka concept seekhte hain.
File mein likha hai:
Support tells us how frequency an item or item combination appears in all transactions
Simple Hinglish mein:
Support batata hai ki koi item ya items ka combination total transactions mein kitni baar aaya.
Maan lo 5 customers ne shopping ki aur 4 customers ne Milk kharida.
Toh Milk ka Support hoga:
\[
\text{Support(Milk)}=\frac{4}{5}=0.8=80\%
\]

Ab formula ki har cheez samjho.
- Support(Milk): Total shopping mein Milk kitni baar aaya.
- 4: Milk kharidne wale transactions ki sankhya.
- 5: Total transactions.
- 0.8: Decimal mein answer.
- 80%: Percentage mein answer.
Tumhari file mein diya formula hai:
Support(X) = Transaction containing (X) / Total Transaction
Iska matlab hai:
\[
\text{Support(X)}=
\frac{\text{X wale transactions}}{\text{Total transactions}}
\]

Yahan X kisi bhi item ko represent karta hai, jaise Milk, Bread ya Butter.
Milk ka Support step-by-step

Total transactions = 5
T1, T2, T3, T4, T5



Milk wale transactions = 4
T1, T2, T4, T5



Calculation
4 ÷ 5 × 100 = 80%

Isliye file mein likha hai:
Support (Milk) = 4/5 = 0.8 => 80%
Iska meaning hai ki hamare 5 transactions mein se 80% transactions mein Milk tha.
Support ka ek aur example
Agar Bread 4 transactions mein aaya:
\[
\text{Support(Bread)}=\frac{4}{5}=80\%
\]

Agar Milk aur Bread saath mein 3 transactions mein aaye:
\[
\text{Support(Milk,Bread)}=\frac{3}{5}=60\%
\]

Dhyan do: jab hum do items ka Support nikalte hain, toh humein un transactions ko count karna hai jinmein dono items saath mein maujood hon.
Yaad rakhne ka shortcut: Support = Kitni baar?

4. Confidence — X ke baad Y kitni baar mila?
Tumhari file mein next heading hai:
### 2.Confidence
Iska matlab hai ki ab hum Confidence seekhenge.
File mein likha hai:
Confidence measures how often Y occurs when X Occurs.
Simple Hinglish mein:
Confidence batata hai ki jab X hota hai, toh uske saath Y kitni baar hota hai.
Example: Milk → Bread
Hum check kar rahe hain ki Milk kharidne wale customers mein se kitne customers Bread bhi kharidte hain.
File ka formula:
Confidence(X->Y) = Support(X,Y)/Support(X)
Mathematically:
\[
\text{Confidence}(X\to Y)=
\frac{\text{Support}(X,Y)}{\text{Support}(X)}
\]

Yahan:
- X = Milk
- Y = Bread
- X → Y = Milk se Bread ka relationship check karna.
- Support(X,Y) = Milk aur Bread saath kitni baar aaye.
- Support(X) = Milk kitni baar aaya.
File ka example solve karte hain
Confidence(Milk -> Bread) = Support(Milk,Bread)/Support(Milk) = 3/4 = 0.75 => 75%
Step-by-step:
1. Total Milk wale transactions = 4.
2. Milk aur Bread dono wale transactions = 3.
3. Ab 3 ko 4 se divide karo.
4. \(3/4=0.75\).
5. Percentage mein \(0.75\times100=75\%\).
Confidence ka visual example
4 customers ne Milk kharida. Unmein se 3 ne Bread bhi kharida.
Milk + Bread


Milk + Bread


Milk + Bread


Milk only



3 ÷ 4 × 100 = 75%
Green = Milk aur Bread dono. Grey = Sirf Milk, Bread nahi.

Isliye Confidence 75% hai.
Iska matlab yeh nahi ki 75% saare customers Bread kharidte hain. Iska matlab hai ki is example mein Milk kharidne wale customers mein se 75% ne Bread bhi kharida.
Yaad rakhne ka shortcut: Confidence = X hone par Y kitni baar hua?

5. Lift — Dono items ka relationship kitna strong hai?
Tumhari file mein next heading hai:
### 3.Lift
Lift humein batata hai ki X aur Y ka combination, Y ke normally hone ki frequency ke comparison mein, kitna common hai.
File mein likha hai:
Lift tells us whether X and Y Occur together more often than we would expected by chance
Simple Hinglish mein:
Lift check karta hai ki X aur Y saath mein expected frequency se zyada baar aa rahe hain, kam baar aa rahe hain, ya normal frequency ke aas-paas hain.
File ka formula:
Lift(X -> Y) = Confidence(X->Y)/Support(Y)
Mathematically:
\[
\text{Lift}(X\to Y)=
\frac{\text{Confidence}(X\to Y)}
{\text{Support}(Y)}
\]

Is formula mein:
- Confidence(X → Y) = X wale transactions mein Y ka percentage.
- Support(Y) = Sabhi transactions mein Y ka percentage.
Milk → Bread ka Lift
File mein diya hai:
Lift(Milk->Bread) = 0.75/0.80 => 0.9375
Hum pehle se jaante hain:
- Confidence(Milk → Bread) = 0.75
- Support(Bread) = 0.80
Ab divide karo:
\[
\text{Lift}=\frac{0.75}{0.80}=0.9375
\]

Answer 0.9375 hai.
Iska meaning hai ki Milk kharidne wale group mein Bread ki frequency, overall Bread ki frequency se thodi kam hai.
Kyun? Kyunki Milk wale transactions mein Bread 75% baar aaya, jabki overall transactions mein Bread 80% baar aaya.
Lift ko kaise interpret karein?
Tumhari file mein yeh table diya gaya hai:
Lift ki value	File mein diya meaning	Simple explanation
Lift < 1	Negative Association	X ke saath Y, expected comparison mein kam hota hai.
Lift = 1	No Association	X ke hone se Y ki frequency mein koi difference nahi dikhta.
Lift > 1	Positive Association	X ke saath Y, expected comparison mein zyada hota hai.
Lift ko example se samjho

Lift = 0.5
Negative association: Y, X wale group mein comparatively kam hota hai.




Lift = 1
No association: Y ki frequency expected frequency ke barabar hai.




Lift = 2
Positive association: Y, X wale group mein expected comparison se zyada hota hai.




Hamare example mein Lift = 0.9375 hai, jo 1 se kam hai. Isliye Milk → Bread ka association thoda negative hai, yaani Bread Milk wale transactions mein overall ke comparison mein thoda kam aaya.
Ek important baat: Lift relationship dikhata hai, lekin yeh prove nahi karta ki Milk kharidne ki wajah se customer Bread kharidta hai.

6. Major Association Rule Algorithms
Tumhari file ki next heading hai:
### Major Association Rule Algorithm
Iska matlab hai ki ab hum Association Rules discover karne ke liye use hone wale main algorithms padhenge.
Algorithm ka simple meaning hai: kisi problem ko solve karne ke liye step-by-step method.
Tumhari file mein do algorithms diye gaye hain:
1. Apriori
2. FP-Growth
6.1 Apriori Algorithm
File mein likha hai:
1. Apriori
Yeh algorithm ka naam hai.
Agli line:
- The classic association rule algorithm
Iska matlab hai ki Apriori ek classic, yaani traditional aur well-known, association rule algorithm hai.
Agli line:
- Apriori -> Find Frequent Itemsets -> Generate association rules
Yeh line Apriori ka workflow samjhati hai.
Step 1: Apriori
Algorithm start hota hai.

Step 2: Find Frequent Itemsets
Aise items ya combinations khojo jo frequently aate hain.

Step 3: Generate Association Rules
Items ke beech rules banao, jaise Milk → Bread.


Frequent Itemset kya hota hai?
Itemset ka matlab hai ek ya zyada items ka group.
- {Milk} = Ek item ka set.
- {Milk, Bread} = Do items ka set.
- {Milk, Bread, Butter} = Teen items ka set.
Frequent ka matlab hai ki items dataset mein kaafi baar aaye hain. Kitni baar aana sufficient hai, yeh minimum support threshold par depend karta hai.
Example: Agar shopkeeper decide karta hai ki kam se kam 40% transactions mein aane wale combinations hi consider karne hain, toh 40% minimum support threshold hoga.
Hamare data mein Milk aur Bread saath 3 out of 5 transactions mein aaye:
\[
\text{Support(Milk,Bread)}=\frac35=60\%
\]

60%, 40% se zyada hai, isliye yeh combination frequent itemset ke liye qualify karta hai.
Apriori ka simple purpose: Frequent combinations find karna aur unse useful rules generate karna.
6.2 FP-Growth Algorithm
File mein likha hai:
2. FP-Growth
Yeh doosre algorithm ka naam hai.
Agli line:
- An optimized approch that uses an FP-Tree
Yahan approch spelling ka intended word approach hai.
Simple Hinglish mein, FP-Growth frequent itemsets discover karne ka ek efficient approach hai. Yeh data ko FP-Tree (Frequent Pattern Tree) naam ki tree structure mein organize karta hai.
Tree ko family tree jaisa samajh sakte ho, jahan related information branches ke form mein arranged hoti hai.
Agli line:
- FP-Growth -> FP-Tree -> Frequent Items -> Association Rules
Iska matlab hai:
Step 1: FP-Growth

Step 2: FP-Tree
Transactions ko tree structure mein organize karo.

Step 3: Frequent Items
Frequently aane wale items aur combinations find karo.

Step 4: Association Rules
Useful relationships generate karo.


Apriori vs FP-Growth
Feature	Apriori	FP-Growth
Purpose	Frequent itemsets find karna	Frequent itemsets find karna
Main approach	Candidate itemsets generate karke check karna	FP-Tree structure use karna
Typical performance	Large datasets par zyada candidate checking ho sakti hai	Aksar zyada efficient hota hai
Result	Frequent patterns aur association rules	Frequent patterns aur association rules
Dono algorithms ka main goal similar hai. FP-Growth ko frequent patterns discover karne ke liye aksar Apriori se zyada efficient banaya gaya hai, khaaskar jab bahut saare candidate combinations hon.

7. File mein HTML tags ka kya meaning hai?
Tumhari notebook mein transactions aur Lift ke tables HTML tags se banaye gaye hain. Yeh Python commands nahi hain; yeh table ko display karne ke formatting instructions hain.
Tag	Simple meaning
<table>	Table start karta hai.
<tr>	Table ki ek row banata hai.
<th>	Heading ya header cell banata hai.
<td>	Normal data cell banata hai.
</table>	Table end karta hai.
</tr>, </th>, </td>	Apne corresponding element ko close karte hain.
<b>...</b>	Text ko bold dikhata hai.
<br>	Nayi line par le jaata hai.
###	Markdown mein heading banata hai.
->	Direction ya relationship dikhata hai, jaise Milk → Bread.
File ki last cell empty hai. Usmein koi additional explanation ya code nahi hai.
8. Complete lecture ka quick revision

Association Rules
Data mein items ke beech relationships discover karna.




Support
Koi item ya combination total transactions mein kitni baar aaya.
Milk = 4/5 = 80%




Confidence
X wale transactions mein Y kitni baar aaya.
Milk → Bread = 3/4 = 75%




Lift
X ke saath Y ki frequency, Y ki overall frequency ke comparison mein kaisi hai.
Milk → Bread = 0.75/0.80 = 0.9375




Algorithms
Apriori aur FP-Growth frequent itemsets find karne ke liye use hote hain.



9. Ab apna knowledge test karo
1. Support kya batata hai?

Item kitni baar total transactions mein aaya

Sirf do items ka price

Model ki accuracy



2. Milk 5 transactions mein se 4 mein aaya. Support kitna hai?

20%

80%

75%



3. Milk 4 transactions mein aaya aur 3 mein Bread bhi tha. Confidence (Milk → Bread) kitna hai?

60%

80%

75%



4. Lift = 1 ka kya matlab hai?

Y ki frequency expected comparison se double hai

X ke saath Y ki frequency, Y ki overall frequency ke barabar hai

X aur Y kabhi saath nahi aate



5. Kaunse do algorithms file mein diye gaye hain?

K-Means aur DBSCAN

Linear Regression aur KNN

Apriori aur FP-Growth

Interview ke liye ek line yaad rakhna: Association Rule Learning discovers relationships between items using measures such as Support, Confidence and Lift. Apriori and FP-Growth are two common algorithms used to find frequent itemsets.
```
