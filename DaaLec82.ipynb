```python
from bs4 import BeautifulSoup
```

```python
soup = BeautifulSoup(content.text , "html.parser")
```

```python
len(soup)
```

```python
topics=[]
desc=[]
links=[]
```

```python
for row in soup.find_all('a', class_='no-underline flex-1 d-flex flex-column'):
    topics.append(row.find('p').text.strip())
    desc.append(row.find_all('p')[1].text.strip())
    links.append('https://github.com' + row['href'])
```

```python
import pandas as pd
```

```python
df=pd.DataFrame({
    'Topic':topics,
    'Description':desc,
    'Link':links
    })
```

```python
df
```

```python
users=[]
repo=[]
links=[]
stars=[]
```

```python
for url in df['Link']:
    content=requests.get(url)
    soup=BeautifulSoup(content.text,'html.parser')
    for row in soup.find_all('div',class_='d-flex flex-justify-between flex-items-start flex-wrap gap-2 tmp-my-3'):
        users.append(row.find('a').text.strip())
        repo.append(row.find_all('a')[1].text.strip())
        stars.append(row.find('span' , id='repo-stars-counter-star')['title'].strip().replace(',',''))
        links.append('https://github.com'+row.find_all('a')[1]['href'])
```

```python
print(len(users))
print(len(repo))
print(len(stars))
print(len(links))
```

```python
df2 = pd.DataFrame({
    'Username': users,
    'Repository': repo,
    'Stars': stars,
    'Link': links
})
```
