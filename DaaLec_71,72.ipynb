linspace() = "Mujhe start aur end ke beech itne equally spaced numbers de do."

Syntax:

np.linspace(start, stop, num)
start = Kahan se start
stop = Kahan tak
num = Kitne numbers chahiye
Example
import numpy as np

print(np.linspace(1, 5, 5))

Output:

[1. 2. 3. 4. 5.]
print(np.linspace(1, 5, 3))

Output:

[1. 3. 5.]
🎯 Memory Trick

np.arange() → Step batao.

np.arange(1, 10, 2)
# 1 3 5 7 9

np.linspace() → Total numbers batao.

np.linspace(1, 10, 5)
# 1.00 3.25 5.50 7.75 10.00
🧠 One-line Formula

linspace() = "Equal gap wale fixed number of values." 🚀

Pandas is an open-source Python library used primarily for data manipulation, cleaning, and analysis. It provides fast, flexible data structures designed to make working with structured (tabular) data intuitive and efficient, acting essentially like Excel or SQL databases inside your code.
Pandas in the Easiest Way (Step by Step)
Step 1: Import Pandas
import pandas as pd

👉 pd = short name of pandas.

Step 2: Two Main Data Structures
1. Series (1 Column)
s = pd.Series([10, 20, 30])
print(s)

Think of it as one Excel column.

2. DataFrame (Rows + Columns)
df = pd.DataFrame({
    "Name": ["Aman", "Rahul"],
    "Age": [20, 22]
})
print(df)

Think of it as an Excel sheet.

6 Main Uses of Pandas
1. Load & Save Data

Read data:

df = pd.read_csv("data.csv")

Save data:

df.to_excel("output.xlsx", index=False)

📌 Remember:
read_* = Load
to_* = Save

2. Clean Data

Remove missing values

df.dropna()

Fill missing values

df.fillna(0)

Remove duplicate rows

df.drop_duplicates()

📌 Remember:

drop = Remove
fill = Fill
3. Filter Data

Age > 25

df[df["Age"] > 25]

Select columns

df[["Name", "Age"]]

📌 Remember:
Condition inside []

4. Group Data

Average salary by department

df.groupby("Department")["Salary"].mean()

📌 Formula:

groupby() → mean()
5. Statistics
df.describe()

Shows:

Count
Mean
Min
Max
Quartiles

📌 One command = Full summary

6. Date & Time

Convert text to date

df["Date"] = pd.to_datetime(df["Date"])

Monthly total

df.resample("M", on="Date").sum()

📌 Remember:

to_datetime() = Convert to date
resample() = Group by time
