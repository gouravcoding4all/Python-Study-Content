Dimension Reduction  

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
from sklearn.datasets import load_iris
```

```python
iris = load_iris()
```

```python
X=iris.data
y=iris.target
```

```python
print("X shape :",X.shape)
print("Classes :",iris.target_names)
print("Features :",iris.feature_names)
```

```python
df= pd.DataFrame(X,columns=iris.feature_names)
```

```python
df.head()
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
X_scaled=scaler.fit_transform(X)
```

```python
X_scaled[:4]
```

```python
#PCA
from sklearn.decomposition import PCA
```

```python
pca =PCA(n_components=2)
```

```python
X_pca=pca.fit_transform(X_scaled)
```

```python
X_pca[:4]
```

```python
df_pca = pd.DataFrame(X_pca,columns=['PCA1','PCA2'])
```

```python
df_pca.head(2)
```

```python
#LDA:Linear Discriminant Analysis
#PCA doesn't use 'y'
#LDA uses 'y' to find maximum class sepration
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
```

```python
LDA = LinearDiscriminantAnalysis(n_components=2)
```

```python
X_LDA =LDA.fit_transform(X_scaled,y)
```

```python
X_LDA.shape
```

```python
1. Dimension Reduction

Notebook ki first heading hai:

Dimension Reduction
Simple meaning

Socho tumhare paas ek school bag hai jisme 100 books hain.

Tumhe sirf important books lekar jaana hai.

Toh tum kya karoge?

👉 Important information rakhoge aur unnecessary cheezein kam karoge.

Machine Learning mein bhi same idea hai.

Agar dataset mein bahut saare columns/features hain:

Feature 1
Feature 2
Feature 3
Feature 4
Feature 5
...
Feature 100

toh hum important information ko kam dimensions/features mein represent karne ki koshish karte hain.

Is process ko Dimension Reduction kehte hain.

Is notebook mein:
4 Features
   ↓
PCA
   ↓
2 Features

Aur LDA bhi:

4 Features
   ↓
LDA
   ↓
2 Features
2. Pandas import
import pandas as pd
Ek-ek part:
import

Python ko bol raha hai:

"Mujhe ek library use karni hai."

pandas

Pandas ek Python library hai jo tables/data handle karne ke kaam aati hai.

Example:

Name	Age	Salary
Rahul	20	30000
Amit	22	40000
as pd

Hum pandas ka short naam pd rakh rahe hain.

Ab baar-baar:

pandas.DataFrame()

likhne ki zarurat nahi.

Hum:

pd.DataFrame()

likh sakte hain.

Simple analogy

Pandas = school ka bada register 📒

pd = us register ka short nickname.

3. Matplotlib import
import matplotlib.pyplot as plt

Matplotlib ka use graphs/charts banane ke liye hota hai.

pyplot ka short name:

plt

Isliye baad mein hum likh sakte hain:

plt.plot()
plt.show()
Simple analogy

Matplotlib = drawing box 🎨

plt = drawing box ka short name.

4. Iris dataset import
from sklearn.datasets import load_iris

Yahan hum Scikit-learn se ek ready-made dataset le rahe hain.

sklearn

Scikit-learn ek popular Machine Learning library hai.

datasets

Iske andar bahut saare practice datasets hote hain.

load_iris

Iris naam ka famous flower dataset load karne ka function hai.

5. Iris dataset load karna
iris = load_iris()

Ye line Iris dataset ko memory mein load karti hai.

Ab:

iris

ke andar dataset ki information available hai.

Isme flowers ke baare mein data hai.

6. X aur y banana
X = iris.data

X mein input/features aa rahe hain.

Iris dataset mein 4 features hain:

Sepal Length
Sepal Width
Petal Length
Petal Width

So:

X = flower ki measurements

Next:

y = iris.target

y mein flower ki class/category hai.

Example:

0 → setosa
1 → versicolor
2 → virginica
Simple analogy

Agar teacher tumse pooche:

"Student ki height, weight aur age kya hai?"

Ye inputs/features hain → X

Aur:

"Student pass hai ya fail?"

Ye target/output hai → y

7. X ka shape check karna
print("X shape :", X.shape)

Output:

X shape : (150, 4)

Iska matlab:

150 = rows
4 = columns

Matlab dataset mein:

150 flowers hain.

Aur har flower ke:

4 features hain.

Table imagine karo:
Flower	Sepal Length	Sepal Width	Petal Length	Petal Width
1	5.1	3.5	1.4	0.2
2	4.9	3.0	1.4	0.2
3	4.7	3.2	1.3	0.2

150 rows × 4 columns.

8. Classes print karna
print("Classes :", iris.target_names)

Output:

Classes : ['setosa' 'versicolor' 'virginica']

Ye flowers ki 3 categories hain.

Simple:
0 → setosa
1 → versicolor
2 → virginica
9. Features print karna
print("Features :", iris.feature_names)

Output:

Features : [
'sepal length (cm)',
'sepal width (cm)',
'petal length (cm)',
'petal width (cm)'
]

Yaani flower ko describe karne ke liye 4 measurements hain.

10. DataFrame banana
df = pd.DataFrame(X, columns=iris.feature_names)

Ye bahut important line hai.

Hum X ko ek proper Pandas table mein convert kar rahe hain.

pd.DataFrame()

Data ko table mein convert karta hai.

X

Hamare flower measurements.

columns=iris.feature_names

Columns ke naam ye honge:

sepal length
sepal width
petal length
petal width
11. First 5 rows dekhna
df.head()

head() normally first 5 rows dikhata hai.

Output mein:

0    5.1    3.5    1.4    0.2
1    4.9    3.0    1.4    0.2
2    4.7    3.2    1.3    0.2
3    4.6    3.1    1.5    0.2
4    5.0    3.6    1.4    0.2
Important

head() ka matlab:

"Mujhe table ke starting ke records dikhao."

12. StandardScaler import
from sklearn.preprocessing import StandardScaler

Ab hum StandardScaler la rahe hain.

Iska kaam hai:

Features ko ek common scale par lana.

Suppose:

Age = 20
Salary = 50000

Salary ka number bahut bada hai.

Machine Learning algorithm mein different scales problem create kar sakte hain.

StandardScaler values ko standard scale par convert karta hai.

13. Scaler object banana
scaler = StandardScaler()

Hum StandardScaler ka ek object bana rahe hain.

Socho:

StandardScaler = machine

Jo data ko standard form mein convert karegi.

14. Data standardize karna
X_scaled = scaler.fit_transform(X)

Ye line bahut important hai.

Isme do kaam hote hain:

fit

Scaler data ko samajhta hai.

Jaise:

"Data ka average kitna hai?"

"Data kitna spread hai?"

transform

Phir data ko standard scale mein convert karta hai.

fit_transform

Dono kaam ek saath:

fit
 ↓
transform
15. First 4 scaled rows
X_scaled[:4]

[:4] ka matlab:

Starting ki 4 rows do.

Output:

[[-0.9006,  1.0190, -1.3402, -1.3154],
 [-1.1430, -0.1320, -1.3402, -1.3154],
 [-1.3853,  0.3284, -1.3970, -1.3154],
 [-1.5065,  0.0982, -1.2833, -1.3154]]

Pehle values thi:

5.1
3.5
1.4
0.2

Ab scaled values ho gayi:

-0.90
1.01
-1.34
-1.31
Don't worry!

Ye negative numbers galat nahi hain.

Standardization ke baad negative values normal hain.

16. PCA

Notebook mein:

#PCA

# ka matlab comment.

Python isko execute nahi karta.

Ye sirf human ko batane ke liye hai:

"Ab PCA start ho raha hai."

17. PCA import
from sklearn.decomposition import PCA

Scikit-learn se PCA import kar rahe hain.

PCA full form:

Principal Component Analysis

PCA ek Dimension Reduction technique hai.

18. PCA object
pca = PCA(n_components=2)

Ye bahut important line hai.

Hum PCA ko bol rahe hain:

"Mere data ke 4 features ko reduce karke 2 dimensions bana do."

Pehle:

4 dimensions

Baad mein:

2 dimensions
Visual:
Sepal Length
Sepal Width
Petal Length
Petal Width
       ↓
      PCA
       ↓
    PCA1
    PCA2
19. PCA apply karna
X_pca = pca.fit_transform(X_scaled)

Yahan PCA:

Data ko samajhta hai
Important patterns find karta hai
4 features ko 2 new features mein convert karta hai
Input:
150 × 4
Output:
150 × 2
20. First 4 PCA rows
X_pca[:4]

Output:

[
[-2.2647,  0.4800],
[-2.0809, -0.6741],
[-2.3642, -0.3419],
[-2.2993, -0.5973]
]

Ab har flower ke paas sirf:

PCA1
PCA2

hai.

Pehle:

4 values

Ab:

2 values
21. PCA DataFrame
df_pca = pd.DataFrame(
    X_pca,
    columns=['PCA1','PCA2']
)

Hum PCA ke result ko Pandas DataFrame mein convert kar rahe hain.

Aur columns ke naam rakh rahe hain:

PCA1
PCA2

So table kuch aisi hogi:

	PCA1	PCA2
0	-2.2647	0.4800
1	-2.0809	-0.6741
2	...	...
22. PCA DataFrame ki first 2 rows
df_pca.head(2)

Yahan:

head(2)

ka matlab:

First 2 rows dikhao.

Output:

       PCA1      PCA2
0 -2.264703  0.480027
1 -2.080961 -0.674134
23. LDA section

Notebook mein:

#LDA:Linear Discriminant Analysis

LDA ka full form:

Linear Discriminant Analysis

Ye bhi Dimension Reduction technique hai.

Lekin PCA aur LDA mein ek important difference hai.

24. PCA vs LDA

Notebook mein likha hai:

#PCA doesn't use 'y'

PCA ko y ki zarurat nahi hoti.

Matlab PCA ko ye nahi pata:

0 = setosa
1 = versicolor
2 = virginica

PCA mainly data ke variation/pattern ko dekhta hai.

Next:

#LDA uses 'y' to find maximum class sepration

LDA y ka use karta hai.

Yaani LDA ko pata hai:

Ye setosa hai
Ye versicolor hai
Ye virginica hai

Phir LDA koshish karta hai ki different classes ko achhe se separate kar sake.

Simple school example

Maan lo class mein 3 groups hain:

Group A
Group B
Group C

PCA bolega:

"Main students ke data ko dekhoonga aur important direction find karunga."

LDA bolega:

"Mujhe pata hai kaun A, B aur C group mein hai. Main aisi direction find karunga jahan groups alag-alag dikh sakein."

25. LDA import
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis

Scikit-learn se LDA import kar rahe hain.

26. LDA object banana
LDA = LinearDiscriminantAnalysis(n_components=2)

Hum LDA ko bol rahe hain:

"Mujhe data ko 2 dimensions mein reduce karna hai."

So:

4 features
   ↓
LDA
   ↓
2 features
27. LDA apply karna
X_LDA = LDA.fit_transform(X_scaled, y)

Ye notebook ki most important lines mein se ek hai.

Yahan do inputs diye gaye hain:

X_scaled

aur

y
X_scaled

Features/data.

y

Classes/labels.

Example:

X_scaled → flower measurements

y → flower ka type
fit_transform

LDA pehle data ko samajhta hai:

fit

Phir usko new dimensions mein convert karta hai:

transform
28. LDA shape
X_LDA.shape

Output:

(150, 2)

Iska matlab:

150 rows
2 columns

Pehle:

X shape = (150, 4)

Ab:

X_LDA shape = (150, 2)

So dimension reduction successfully hua.

🧠 Ab poori notebook ko ek story se samjho

Imagine karo tumhare paas 150 flowers hain.

Har flower ke paas 4 measurements hain:

1. Sepal Length
2. Sepal Width
3. Petal Length
4. Petal Width

So:

150 flowers × 4 measurements
Step 1 — Data load
iris = load_iris()

Flowers ka data le aaye.

Step 2 — X aur y
X = iris.data
y = iris.target
X = measurements
y = flower category
Step 3 — Data ko standardize
X_scaled = scaler.fit_transform(X)

Sab measurements ko comparable scale par le aaye.

Step 4 — PCA
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)
4 features
    ↓
   PCA
    ↓
2 features

PCA y ko use nahi karta.

Step 5 — LDA
LDA = LinearDiscriminantAnalysis(n_components=2)

X_LDA = LDA.fit_transform(X_scaled,y)
4 features
    ↓
   LDA
    ↓
2 features

LDA y ko use karta hai taaki classes ko separate karne mein help mile.

⭐ PCA vs LDA — 5th Class Version
PCA	LDA
Dimension Reduction	Dimension Reduction
y use nahi karta	y use karta hai
Data ke important patterns dekhta hai	Classes ko separate karne par focus karta hai
Unsupervised technique	Supervised technique
X sufficient hai	X + y chahiye
Example: important information bachana	Example: different teams ko alag karna
🔥 Sabse important concept

Tumhari notebook ka main flow yaad rakho:

             IRIS DATASET
                  ↓
             X = Features
             y = Classes
                  ↓
          StandardScaler
                  ↓
             X_scaled
             /       \
            /         \
          PCA         LDA
           ↓           ↓
        PCA1,PCA2   LDA1,LDA2
           ↓           ↓
        4 → 2       4 → 2
One-line interview answer:

Dimension Reduction means dataset ke bahut saare features ko kam dimensions mein convert karna while trying to preserve important information.

PCA data ke important variation/pattern ko use karke dimensions reduce karta hai, while LDA class labels y ka use karke classes ko better separate karne wali dimensions find karta hai.
```
