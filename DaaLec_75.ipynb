```python
#Data Transformation/Column Transformation/Feature Engneering
```

```python
#347K,23.4K,2M,1.1M,373
```

```python
def convertVotes(val):
    val=str(val)
    if val[-1]=='K':
        return int(float(val[0:len(val)-1])*1000)
    elif val[-1]=='M':
        return int(float(val[0:len(val)-1])*1000000)
    else:
        return int(val)
```

```python
convertVotes('23')
```

```python
df.Votes=df.Votes.map(convertVotes)
```

```python
df.Votes
```

map() ka use 👶

map() = column ki har value par ek function apply karna.

df.Votes = df.Votes.map(convertVotes)

➡️ Votes ki har value convertVotes function mein jayegi.

Example:

Votes
1K
2K
5K

map(convertVotes) → har value ko convert karega:

1000
2000
5000

🧠 Short trick:

map() = Har value par function lagao. ✅
