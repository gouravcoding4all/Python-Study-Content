```python
def convertDuration(val):
    if len(val.split())==2:
        h , m = tuple(val.split())
        return int(h[0: len(h)-1 ])*60+int(m[0: len(m)-1 ])
    elif val[-1]=='h':
        return int(val[0: len(val)-1 ])*60
    elif val[-1]=='m':
        return int(val[0: len(val)-1 ])
    else:
        return convertDuration(df.Duration.mode()[0])
```

def convertDuration(val):

➡️ Duration ko minutes mein convert karne ka function.

if len(val.split()) == 2:

➡️ Agar value 1h 30m jaisi hai.

h, m = tuple(val.split())

➡️ 1h aur 30m ko alag karo.

return int(h[0:len(h)-1])*60 + int(m[0:len(m)-1])

➡️ Hours × 60 + Minutes = total minutes.

elif val[-1] == 'h':

➡️ Sirf hours hain → hours × 60.

elif val[-1] == 'm':

➡️ Sirf minutes hain → same minutes.

else:
    return convertDuration(df.Duration.mode()[0])

➡️ Format galat/unknown hai → most common Duration lekar convert karo.

🧠 One-line Trick:

h → ×60 | m → same | h + m → ×60 + m | गलत → Mode ✅
