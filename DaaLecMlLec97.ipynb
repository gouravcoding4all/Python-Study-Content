```python
import seaborn as sns
```

```python
from sklearn.datasets import fetch_openml
```

```python
titanic = fetch_openml("titanic")
```

```python
df = titanic.frame
```

```python
#df=sns.load_dataset('titanic')
```

```python
df.shape
```

```python
df.head(2)
```

```python
from sklearn.cluster import KMeans
```

```python
kmeans=KMeans(n_clusters=2)
```

```python
labels = kmeans.fit_predict(df[['Age','Class']])
```

```python
labels
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(6,3))
plt.scatter(df['Age'],df['Class'])
plt.show()
```

```python
plt.figure(figsize=(6,3))
plt.scatter(df['Age'],df['Class'],c=labels)
plt.show()
```
