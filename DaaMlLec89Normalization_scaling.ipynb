```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

plt.plot(), plt.bar(), plt.pie(), plt.scatter() etc. isi pyplot module ke functions hain.

```python
dataset=pd.read_excel(r"C:\Users\dell\Downloads\1775106859-Financial_Sample.xlsx")
dataset.head(4)
```

```python
dataset.isnull().sum()
```

```python
dataset.select_dtypes(include='object').columns
```

```python
for col in ['Segment', 'Country', 'Product', 'Discount Band', 'Month Name']:
    dataset[col] = dataset[col].fillna(dataset[col].mode()[0])
```

```python
dataset.isnull().sum()
```

```python
dataset.select_dtypes(include=['int', 'float']).columns
```

```python
for col in ['Units Sold', 'Manufacturing Price', 'Sale Price', 'Gross Sales',
       'Discounts', ' Sales', 'COGS', 'Profit', 'Month Number', 'Year']:
    dataset[col] = dataset[col].fillna(dataset[col].mean())
```

```python
dataset.describe()
```

```python
print(dataset.columns.tolist())
```

```python
dataset.columns = dataset.columns.str.strip()
```

```python
sns.histplot(dataset['Gross Sales'], kde=True)
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
ms=MinMaxScaler()
```

```python
ms.fit_transform(dataset[["Gross Sales"]])
```

```python
dataset['Gross Sales_min']=ms.fit_transform(dataset[["Gross Sales"]])
dataset.head()
```
